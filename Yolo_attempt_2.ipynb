```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("agrigorev/clothing-dataset-full")

print("Path to dataset files:", path)
```

```python
import os
import shutil

# Define the new directory name
destination_dir = 'cloths'

# Create the new directory if it doesn't exist
if not os.path.exists(destination_dir):
    os.makedirs(destination_dir)
    print(f"Created directory: {destination_dir}")

# Copy the contents of the downloaded dataset to the new directory
# shutil.copytree requires the destination directory to not exist, so we'll copy contents directly
for item in os.listdir(path):
    s = os.path.join(path, item)
    d = os.path.join(destination_dir, item)
    if os.path.isdir(s):
        # Use dirs_exist_ok=True to allow merging if directory already exists
        shutil.copytree(s, d, symlinks=False, ignore=None, dirs_exist_ok=True)
    else:
        shutil.copy2(s, d)

print(f"Contents copied to: {destination_dir}")
print(f"Files in '{destination_dir}': {os.listdir(destination_dir)}")
```

```python
import os
import shutil
import yaml
import random

# --- Helper function for directory exploration ---
def explore_directory(path, depth=2):
    print(f"Exploring: {path}")
    if not os.path.exists(path):
        print(f"Path does not exist: {path}")
        return
    for root, dirs, files in os.walk(path):
        current_depth = root.count(os.sep) - path.count(os.sep)
        if current_depth > depth:
            continue
        indent = "    " * current_depth
        print(f"{indent}├── {os.path.basename(root)}/")
        # Print directories
        for d in sorted(dirs):
            print(f"{indent}│   ├── {d}/")
        # Print files (limited)
        for i, f in enumerate(sorted(files)):
            if i < 5: # Limit file listing to first 5
                print(f"{indent}│   └── {f}")
        if len(files) > 5:
            print(f"{indent}│   └── ... ({len(files) - 5} more files)")

# --- 1. Check file structures ---
print("--- Exploring downloaded dataset structures ---")

# Explore the 'cloths' dataset
print("\n--- Exploring 'cloths' (Clothing Dataset Full) ---")
# The actual images are within subdirectories
explore_directory('cloths', depth=2)

# Explore the 'yolo_highvis_person_detection_dataset'
print("\n--- Exploring 'yolo_highvis_person_detection_dataset' (YOLO-HiVis and Person Detection Dataset) ---")
explore_directory('yolo_highvis_person_detection_dataset/YOLO-HiVis-Data', depth=2)

# --- 2. Analyze 'cloths' dataset compatibility for YOLO ---
print("\n--- Analyzing 'cloths' dataset for YOLO compatibility ---")
# The 'agrigorev/clothing-dataset-full' dataset typically contains images for classification
# and an `images.csv` file with classification labels, not bounding box annotations in YOLO format.
# Therefore, it cannot be directly 'merged' for YOLO object detection training without
# converting its labels (if bounding boxes are implicit/available) or re-annotating.
# For this task, we will focus on the 'yolo_highvis_person_detection_dataset' which is already in YOLO format.
print("The 'cloths' dataset is primarily for classification and does not contain YOLO-formatted bounding box labels. It will not be directly used for YOLO training without label conversion.")

# --- 3. Prepare and Merge 'yolo_highvis_person_detection_dataset' into the 'dataset' structure ---
print("\n--- Preparing 'yolo_highvis_person_detection_dataset' for YOLO training ---")

source_images_dir = 'yolo_highvis_person_detection_dataset/YOLO-HiVis-Data/images'
source_labels_dir = 'yolo_highvis_person_detection_dataset/YOLO-HiVis-Data/labels'

target_train_images_dir = 'dataset/images/train'
target_val_images_dir = 'dataset/images/val'
target_train_labels_dir = 'dataset/labels/train'
target_val_labels_dir = 'dataset/labels/val'

# Ensure target directories exist (already created by previous cell, but good practice)
os.makedirs(target_train_images_dir, exist_ok=True)
os.makedirs(target_val_images_dir, exist_ok=True)
os.makedirs(target_train_labels_dir, exist_ok=True)
os.makedirs(target_val_labels_dir, exist_ok=True)


# Get all image files
image_files = [f for f in os.listdir(source_images_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp', '.tiff'))]
random.shuffle(image_files)

# Define split ratio
train_split = 0.8
num_train = int(len(image_files) * train_split)

train_files = image_files[:num_train]
val_files = image_files[num_train:]

all_class_ids = set()

# Copy training files
print(f"Copying {len(train_files)} training files...")
for img_file in train_files:
    # Copy image
    shutil.copy(os.path.join(source_images_dir, img_file), os.path.join(target_train_images_dir, img_file))

    # Copy corresponding label file
    base_name = os.path.splitext(img_file)[0]
    label_file_src = os.path.join(source_labels_dir, base_name + '.txt')
    if os.path.exists(label_file_src):
        shutil.copy(label_file_src, os.path.join(target_train_labels_dir, base_name + '.txt'))
        # Extract class IDs
        with open(label_file_src, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if parts:
                    all_class_ids.add(int(parts[0]))
    else:
        print(f"Warning: Label file not found for {img_file}")

# Copy validation files
print(f"Copying {len(val_files)} validation files...")
for img_file in val_files:
    # Copy image
    shutil.copy(os.path.join(source_images_dir, img_file), os.path.join(target_val_images_dir, img_file))

    # Copy corresponding label file
    base_name = os.path.splitext(img_file)[0]
    label_file_src = os.path.join(source_labels_dir, base_name + '.txt')
    if os.path.exists(label_file_src):
        shutil.copy(label_file_src, os.path.join(target_val_labels_dir, base_name + '.txt'))
        # Extract class IDs (also for validation set to ensure all classes are captured)
        with open(label_file_src, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if parts:
                    all_class_ids.add(int(parts[0]))
    else:
        print(f"Warning: Label file not found for {img_file}")

print(f"Finished copying files. Total images: {len(image_files)} (Train: {len(train_files)}, Val: {len(val_files)})")

# --- 4. Identify classes and create YAML ---
print("\n--- Generating data.yaml file ---")

# Sort class IDs and create a names list.
# Based on the dataset name, it's very likely class 0 is 'person' and class 1 is 'highvis_person'.
# If more classes are found, a generic naming will be used.
sorted_class_ids = sorted(list(all_class_ids))
class_names = []
if sorted_class_ids == [0, 1]:
    class_names = ['person', 'highvis_person']
elif len(sorted_class_ids) > 0:
    class_names = [f'class_{i}' for i in sorted_class_ids]
else:
    print("Warning: No class IDs found in label files. Assuming default classes or check dataset.")
    class_names = ['object'] # Default if no classes are found

print(f"Detected classes (IDs): {sorted_class_ids}")
print(f"Assigned class names: {class_names}")


data_yaml_content = {
    'path': 'dataset',  # Path to the root of the dataset folder
    'train': 'images/train',
    'val': 'images/val',
    'nc': len(class_names),
    'names': class_names
}

# Convert to YAML string
yaml_string = yaml.dump(data_yaml_content, default_flow_style=False, sort_keys=False)

# Save the YAML file
with open('data.yaml', 'w') as f:
    f.write(yaml_string)

print("\n--- data.yaml created successfully: ---")
print(yaml_string)

print("\n--- Next Steps ---")
print("You can now use this 'data.yaml' with your YOLO model for training, for example:")
print("model.train(data='data.yaml', epochs=100, imgsz=640)")
```
